```python
class Coordinate(NamedTuple):
    lat: float
    lon: float

    def __str__(self):
        ns = 'N' if self.lat >= 0 else 'S'
        we = 'E' if self.lon >= 0 else 'W'
        return f'{abs(self.lat):.1f}°{ns}, {abs(self.lon):.1f}°{we}'
```

```python
rome = Coordinate(42.1, 12.5)
rome
```

```python
rome == Coordinate(42.1, 12.5)
```

```python
rome.lat = 42.0
```

## Using a decorator
The attributes may be set to be mutable

```python
from dataclasses import dataclass, field

@dataclass(frozen=False)
class Coordinate:
    lat: float
    lon: float

    def __str__(self):
        ns = 'N' if self.lat >= 0 else 'S'
        we = 'E' if self.lon >= 0 else 'W'
        return f'{abs(self.lat):.1f}°{ns}, {abs(self.lon):.1f}°{we}'
```

```python
rome = Coordinate(42.1, 12.5)
rome
```

```python
print(rome)
```

```python
rome.lat = 42.0
```

```python
print(rome)
```

```python
@dataclass
class ClubMember:
    name: str
    guests: list = field(default_factory=list)
    athlete: bool = field(default=False, repr=False)
```

We can define a subclass of Classmember with a post initialization function that is called soon after the class is instantiated. 

```python
@dataclass
class HackerClubMember(ClubMember):
    all_handles = set()
    handle: str = ''
    def __post_init__(self):
        cls = self.__class__
        if self.handle == '':
            self.handle = self.name.split()[0]
        if self.handle in cls.all_handles:
            msg = f'handle {self.handle!r} already exists.'
            raise ValueError(msg)
        cls.all_handles.add(self.handle)
```

```python
luigi = HackerClubMember('Luigi Selmi', handle='lgslm_2')
luigi
```

```python
luigi.athlete
```

## Pattern matching
We might look for certain instances of a class with specific attribute values

```python
import typing
class City(typing.NamedTuple):
    continent: str
    name: str
    country: str
```

Let's create a list of objects of type City

```python
cities = [
    City('Asia', 'Tokyo', 'JP'),
    City('Asia', 'Delhi', 'IN'),
    City('North America', 'Mexico City', 'MX'),
    City('North America', 'New York', 'US'),
    City('South America', 'São Paulo', 'BR'),
]
```

and define a function to look for specific object of type City whith attribute *continent='Asia'*

```python
def match_asian_cities():
    results = []
    for city in cities:
        match city:
            case City(continent='Asia'):
                results.append(city)
    return results
```

```python
match_asian_cities()
```

```python
rome = City('Europe', 'Rome', 'IT')
rome
```
