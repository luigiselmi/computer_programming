# Functions as first-class objects
In Python a function is a class object. A function is a *callable object*,  that is, it implements the \_call()__ method. 

```python
def factorial(n):
    '''
    This function calculates the factorial of an integer 
    passed as the arguument.
    '''
    return 1 if n < 2 else n * factorial(n - 1)
```

```python
factorial(10)
```

The \__doc__ special method returns the description of the function, if available.

```python
factorial.__doc__
```

```python
type(factorial)
```

Since a function is a class object it can be bound to a variable

```python
fact = factorial
```

```python
fact(3)
```

## Passing a function as argument to another function
A function can be passed to other functions. For example we pass the factorial to the map() function that maps the result of factorial to an iterable, e.g. range(11)

```python
list(map(factorial, range(11)))
```

A function that takes or returns another function is a 2nd order, or higher order, function. the sorted() function can take a function, e.g. len(), as an argument to sort a list of words by their length.

```python
fruits = ['strawberry', 'fig', 'apple', 'cherry', 'raspberry', 'banana']
```

```python
sorted(fruits, key=len)
```

## map(), filter() and reduce()
These functions are used in functional programming and are also available in Python but there are better options. **List comprehensions** can be a substitute for the map() function 

```python
[factorial(i) for i in range(11) ]
```

A lst comprehension can also replace a filter

```python
[factorial(n) for n in range(6) if n % 2]
```

The reduce() function, used to add a sequence, can be replaced by the sum() function 

```python
sum(range(5))
```

# all() and any()
These are two other reducing built-in function

```python
import numpy as np
a = np.array([1, 2, 3, 4])
```

```python
all(a > 0)
```

```python
any(a > 0)
```

```python
b = np.array([1, 2, 3, -1])
```

```python
all(b > 0)
```

```python
any(b > 0)
```
