# How to reference a module up in the hierarchy
We might want to import functions defined in modules up in the hierarchy

```python
import sys
from pathlib import Path

# 1. Trova la cartella del notebook attuale
notebook_dir = Path().resolve()

# 2. Risali alla cartella superiore (la radice del progetto)
# Aggiungi altri .parent se la radice si trova più in alto di un solo livello
project_root = notebook_dir.parent
print(project_root)
```

```python
# 3. Aggiungi la radice a sys.path se non è già presente
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
```

```python
# 4. Ora puoi importare liberamente le tue funzioni
# (Sostituisci "nome_cartella_superiore" con il nome reale della directory)
from fibo import fib
```
