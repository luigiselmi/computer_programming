# Modules
Modules are used to separate the namespaces in Python. Two functions can be defined using the same name if they belong to two different namespaces. Here we see how modules, constants, functions and classes can be imported in a Python script or Jupyter notebook. We follow the Python documentation in section [6. Modules](https://docs.python.org/3/tutorial/modules.html). 

The script *fibo.py* is saved in the same notebook's folder.

```python
import fibo
```

We can call the functions defined in the module

```python
fibo.fib(100)
```

We can import the module's functions directly

```python
from fibo import fib, fib2
```

```python
fib(100)
```

When a module, e.g. *fibo*, is imported the interpreter searches for a built-in module, then for a file named *fibo.py* in al list of directories listed in the variable *sys.path*

```python
import sys
len(sys.path)
```

A new path can be added

```python
sys.path.append('/ufs/guido/lib/python')
len(sys.path)
```

We can use the built-in function *dir()* to see the new names added to the path by the *fibo* module

```python
dir(fibo)
```

We can see the list of built-in functions

```python
import builtins
dir(builtins)[:10]
```

## Packages
A package is a collection of modules within the same folder. Modules can be organized in a hierarchy of packages. Modules in subfolder are named using the dot convenction. A package should contain a special module *\__init__.py* in order for the Python interpreter to treat that directory as a package. The *\__init.py__* file can be empty or contain some initialialization code. Importing a module in a subdirectory is straightforward as shown in the example.

```python
from mysubfolder.math_functions import add, Point
```

```python
add(1, 2)
```

```python
p1 = Point(2.0, 3.0)
print(p1)
```

```python
p1.length()
```
