# The Python data model
The Python data model implements standard data structures, such as collection types, and algorithms that can be used in any Python script without re-inventing them. One such data structure is the *collections*. Collections can be sequences, sets, and mappings (i.e. dictionaries)

A class definition with a constructor, attributes and methods. A french deck is a collection of 52 cards that implements methods to count the number of items and pick up an item by its index. The class is immutable. 

```python
class FrenchDeck:
    ranks = [str(n) for n in range(2, 11)] + list('JQKA')
    suits = 'spades diamonds clubs hearts'.split()
    
    def __init__(self):
        self._cards = [Card(rank, suit) for suit in self.suits for rank in self.ranks]

    def __len__(self):
        return len(self._cards)

    def __getitem__(self, position):
        return self._cards[position]
```

```python
deck = FrenchDeck()
```

```python
len(deck)
```

We can pick up one item at random

```python
from random import choice
choice(deck)
```

```python
deck[0:12]
```

```python
for card in deck:
    print(card)
```

## A Vector class
A class that implements some special methods

```python
import math
class Vector:
    
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f'Vector({self.x!r}, {self.y!r})'
    
    def __abs__(self):
        return math.hypot(self.x, self.y)
    
    def __bool__(self):
        return bool(abs(self))
    
    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Vector(x, y)
    
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)
```

```python
v1 = Vector(2, 4)
v1
```

```python
v2 = Vector(2, 1)
v2
```

The Python interpreter maps the "+" operator to the special method \__add__ defined in the class 

```python
v1 + v2
```

The adition commutative

```python
v2 + v1
```

```python
v1 * 3
```

The multiplication between a vector and a scalar defined in the Vector class is not commutative

```python
3 * v1
```

The *print()* method calls the \__repr__ special method to provide a representation of an object of the Vector class like the one used in the code
