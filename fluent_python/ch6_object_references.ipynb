# Object references
In Python an instance of a class is handled by reference. This can be seen from this example. The variable a is an handle to the list object. If another variable is declared, and passed a reference to the list, when a change is made it is seen from all its handles. 

```python
a = [1, 2, 3]
```

```python
b = a
b
```

```python
a.append(4)
```

```python
b
```

## Object identity and equality
In the 1st example a dictionary object is assigned, i.e bounded, to two variables

```python
charles = {'name': 'Charles L. Dodgson', 'born': 1832}
id(charles)
```

```python
lewis = charles
id(lewis)
```

The *is* operator compares the list object identifier bound to the two variables. The comparison of two instances of the same class is implemented in the \__eq__ special method and for a dictionary object the method simply compares the ids of the instances. Since it's the same the two variable represent the same object. In other words the two variables are **aliases**, that is they are bound to the same object.

```python
lewis is charles
```

In this second example the 2nd object looks the same as the first one, same type and attributes

```python
lewis = {'name': 'Charles L. Dodgson', 'born': 1832}
```

Since the ids of the two objects are dfferent, the *is* operator returns False

```python
lewis is not charles
```

The *==* operator checks whether the objects' attributes are the same or not. Since the two distinct objects have the same description they are meant to be **equal** even if **not identical**

```python
lewis == charles
```

## Function parameters as references
A function parameters are references to the objects passed as arguments to the function. This implies that if the object is mutable and is changed using its reference in the function the change will be seen outside of the function, that is after it has been called. This outcome might not be what is required and is usually a best practise to not change an object that has been received as an argument. One way to make unmutable an object is to copy its data into a data class with the *Frozen* attribute set to True and pass it to to function instead of the original object.

```python
def add(a, b):
    a += b
    return a
```

```python
a = [1, 2]
b = [3, 4]
```

```python
add(a, b)
```

We can see that a is changed

```python
a
```

```python
b
```

## del and garbage collection
An object is removed form memory when its last reference is deleted

```python
a = [1, 2]
b = a
```

```python
del a
```

The object is not deleted since there is still a reference to it

```python
b
```

Rebinding the last reference to a different object removes the last reference of the original one so that the object can be removed from memory.

b = [3]
