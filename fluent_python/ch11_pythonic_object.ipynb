# A Pythonic Object
We use a 2D vector as an example of how to implement a Pyhon class object. A 2D vector is defined by a pair of real values. A Python class object implements several method to check the parameters type, the object string representation and format, its transformation to other formats, its comparison method and many other.

```python
from array import array
import math

class Vector2D:
    typecode = 'd'

    def __init__(self, x, y):
        self.x = float(x)
        self.y = float(y)
    
    def __iter__(self):
        return (i for i in (self.x, self.y))

    def __repr__(self):
        class_name = type(self).__name__
        return '{}({!r}, {!r})'.format(class_name, *self)

    def __str__(self):
        return str(tuple(self))

    def __format__(self, fmt_spec=''):
        components = (format(c, fmt_spec) for c in self)
        return '({}, {})'.format(*components)

    def __bytes__(self):
        return (bytes([ord(self.typecode)]) + bytes(array(self.typecode, self)))

    def __eq__(self, other):
        return tuple(self) == tuple(other)

    def __abs__(self):
        return math.hypot(self.x, self.y)

    def __bool__(self):
        return bool(abs(self))
```

```python
v1 = Vector2D(3, 4)
v1
```

## A set of vectors
A set is a collection of distinct item. In order to build a set each item must have a distinct identity. This can be achieved by implementing a special hash method \__hash()__ for the type of object that returns a value, unlikely to be equal to a previous or future one, for example using its attributes. The object's hash value will be used by the \__eq()__ special method to compare two objects of the same type. We also want the object attributes to be immutable. So far a Vector2D object cannot be used to build a set because it doesn't implement the \__hash()__ method.

```python
hash(v1)
```

We have to add some methods to the Vector2D class to have hashable Vector2D objects

```python
class Vector2D:

    __match_args__ = ('x', 'y') # specify the order of the object's attributes
    
    typecode = 'd'

    def __init__(self, x, y):
        self.__x = float(x)  # the double underscore "__" make the attribute private, i.e. immutable
        self.__y = float(y)

    @property # getter method
    def x(self):
        return self.__x

    @property # getter method
    def y(self):
        return self.__y
    
    def __hash__(self):
        return hash((self.x, self.y))
    
    def __iter__(self):
        return (i for i in (self.x, self.y))

    def __repr__(self):
        class_name = type(self).__name__
        return '{}({!r}, {!r})'.format(class_name, *self)

    def __str__(self):
        return str(tuple(self))

    def __format__(self, fmt_spec=''):
        components = (format(c, fmt_spec) for c in self)
        return '({}, {})'.format(*components)

    def __bytes__(self):
        return (bytes([ord(self.typecode)]) + bytes(array(self.typecode, self)))

    def __eq__(self, other):
        return tuple(self) == tuple(other)

    def __abs__(self):
        return math.hypot(self.x, self.y)

    def __bool__(self):
        return bool(abs(self))
```

```python
v1 = Vector2D(3, 4)
```
